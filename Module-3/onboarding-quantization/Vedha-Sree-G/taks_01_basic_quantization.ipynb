# Tensor Quantization and Dequantization

### Import Required Libraries

```python
import torch
import numpy as np
```

### Quantization

```python
def quantize_tensor(tensor, scale, zero_point):
    scaled_and_shifted_tensor = tensor / scale + zero_point
    rounded_tensor = torch.round(scaled_and_shifted_tensor)
    q_min = torch.iinfo(torch.int8).min
    q_max = torch.iinfo(torch.int8).max
    q_tensor = rounded_tensor.clamp(q_min,q_max).to(torch.int8)
    return q_tensor
```

### Dequantization

```python
def dequantize_tensor(quantized_tensor, scale, zero_point):
    dequantized_tensor = scale * (quantized_tensor.float() - zero_point)
    return dequantized_tensor
```

### Input

```python
tensor = np.array([1.0, -0.5, 3.2, -2.8, 0.0], dtype=np.float32)
tensor = torch.tensor(tensor)
scale =  0.025
zero_point =  0
```

### Perform Quantization & Dequantization

```python
q_tensor = quantize_tensor(tensor,scale,zero_point)
dq_tensor = dequantize_tensor(q_tensor,scale,zero_point)
```

### Compute Error

```python
abs_error = torch.abs(tensor - dq_tensor)
mean_abs_error = torch.mean(abs_error)
max_abs_error = torch.max(abs_error)
```

### Display the Result

```python
print("Original tensor: ",tensor)
print("Quantized tensor (INT8): ", q_tensor)
print("Dequantized tensor (float): ", dq_tensor)
print("Absolute error per element: ", abs_error)
print("Mean Absolute Error (MAE): ", mean_abs_error)
print("Maximum Absolute Error: ", max_abs_error)
```
