# Geometric Transformations

### Import the Required Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Pick a Reference Number

```python
referenceNo = 56
```

### Compute Angle & Scale

```python
angle = referenceNo * 3
scale = 1 + (referenceNo % 10) / 10
```

```python
print("Reference Number:", referenceNo)
print("Rotation Angle:", angle, "degrees")
print("Scale Factor:", scale)
```

### Read Image

```python
image = cv2.imread("Documents/AIML-Assessment/Module-2/raw_captures/Capture_3.jpeg")
image_rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
height, width = image.shape[:2]
```

### Translation

```python
tx = 100
ty = 200
Tmatrix = np.float32([[1,0,tx],
                     [0,1,ty]])
translated_image = cv2.warpAffine(image_rgb, Tmatrix, (width,height))
```

### Rotation & Scaling

```python
center = (width//2,height//2)
Rmatrix = cv2.getRotationMatrix2D(center, angle, scale)
rotated_image = cv2.warpAffine(image_rgb, Rmatrix, (width,height))
```

### Affine Transform

```python
pts1 = np.float32([[50, 50],
                   [200, 50],
                   [50, 200]])

pts2 = np.float32([[10, 100],
                   [200, 50],
                   [100, 250]])
Amatrix = cv2.getAffineTransform(pts1, pts2)
affine_image = cv2.warpAffine(image_rgb, Amatrix, (width,height))
```

### Display the Results

```python
fig, axes = plt.subplots(1,4,figsize=(20,5))
axes[0].imshow(image_rgb)
axes[0].set_title("Original")
axes[0].axis('off')
axes[1].imshow(translated_image)
axes[1].set_title("Translated")
axes[1].axis('off')
axes[2].imshow(rotated_image)
axes[2].set_title("Rotated & Scaled")
axes[2].axis('off')
axes[3].imshow(affine_image)
axes[3].set_title("Affine Transform")
axes[3].axis('off')
```
