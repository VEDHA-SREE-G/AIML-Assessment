# Color Space Conversion

### Import Required Libraries

```python
import cv2
import matplotlib.pyplot as plt
```

### Read Image

```python
img = cv2.imread("Documents/AIML-Assessment/Module-2/raw_captures/Capture_2.jpeg")
```

### Converting to Different Color Spaces

```python
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img_gray = cv2.cvtColor(img_rgb,cv2.COLOR_RGB2GRAY)
img_hsv = cv2.cvtColor(img_rgb,cv2.COLOR_RGB2HSV)
```

### Display the Result

```python
fig, axes = plt.subplots(1,3, figsize=(15,5))
axes[0].imshow(img_gray,cmap="gray")
axes[0].set_title("Gray Scale")
axes[1].imshow(img_hsv)
axes[1].set_title("HSV")
axes[2].imshow(img_rgb)
axes[2].set_title("RGB")
```

### HSV Channel

```python
h, s, v = cv2.split(img_hsv)
fig, axes = plt.subplots(1,3,figsize=(15,5))
axes[0].imshow(h,cmap="gray")
axes[0].set_title("Hue")
axes[1].imshow(s,cmap="gray")
axes[1].set_title("Saturation")
axes[2].imshow(v,cmap="gray")
axes[2].set_title("Value")
```

The V channel represents the brightness of each pixel, so facial features, clothing, and background objects remain clearly visible even under low-light conditions. The Hue channel loses significant detail and appears noisy, while the Saturation channel contains less recognizable information. Therefore, for this particular low-light image, the Value channel provides the best representation of shadow details.
