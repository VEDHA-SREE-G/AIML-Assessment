# Thresholding & Morphology

### Import Required Libraries

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

### Load Image

```python
image = cv2.imread("Documents/AIML-Assessment/Module-2/raw_captures/Capture_4.jpeg",cv2.IMREAD_GRAYSCALE)
plt.imshow(image, cmap="gray")
```

### Apply Otsu's Thresholding

```python
_,otsu = cv2.threshold(image,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
plt.imshow(otsu, cmap="gray")
```

### Use Morphological Closing

```python
kernel = np.ones((2,2), np.uint8)
closed = cv2.morphologyEx(otsu,cv2.MORPH_CLOSE,kernel)
plt.imshow(closed, cmap="gray")
```

### Crop of One Word

```python
x, y, w, h = 330,700,250,80
word_before = otsu[y:y+h, x:x+w]
word_after = closed[y:y+h, x:x+w]
fig, axes = plt.subplots(1,2,figsize=(10,5))
axes[0].set_title("Word Before")
axes[0].imshow(word_before, cmap="gray")
axes[1].set_title("Word After")
axes[1].imshow(word_after, cmap="gray")
```

### Note

Larger kernels produced excessive thickening of the handwriting and caused neighboring strokes to merge, reducing readability. Because the image resolution is 1076x1600 pixels and the handwritten strokes are only a few pixels wide, a 2×2 kernel provided the best balance between gap filling and detail preservation.
