# Real-Time Video Processing

### Import Required Libraries

```python
import cv2
```

### Input Video

```python
cap = cv2.VideoCapture("Documents/AIML-Assessment/Module-2/raw_captures/video.mp4")
```

###  Get Video Properties

```python
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
```

### Output Video Writer

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out1 = cv2.VideoWriter("canny_edge_task9.mp4",fourcc,fps,(width,height))
out2 = cv2.VideoWriter("face_detect_task9.mp4",fourcc,fps,(width,height))
```

### Load Haar Cascade

```python
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
```

###  Video Processing

```python
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    #Canny Edge Detection
    edges = cv2.Canny(frame_gray, 50, 150)
    edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
    out1.write(edges_bgr)
    cv2.imshow("Edge Detection", edges_bgr)
    #Face Detection
    faces = face_cascade.detectMultiScale(frame_gray,scaleFactor=1.2,minNeighbors=5,minSize=(30, 30))
    for (x,y,w,h) in faces:
        cv2.rectangle(frame, (x,y), (x+w, y+h), (0,255,0), 2)
    out2.write(frame)
    cv2.imshow("Face Detection", frame)
    if(cv2.waitKey(1) & 0xFF == ord('q')):
        break
```

```python
cap.release()
out1.release()
out2.release()
cv2.destroyAllWindows()
```

### Bug-Fix

##### Bug-1:
The output video was not saving correctly because cv2.Canny() produced a single-channel grayscale image while VideoWriter expected a 3-channel frame.
##### Fix-1: 
Converted the edge image to BGR using: edges_bgr = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)

##### Bug-2:
During face detection, the Haar Cascade classifier incorrectly detected non-face objects as faces.

##### Fix-2:
Reduced false detections by increasing minNeighbors and setting a minimum face size
